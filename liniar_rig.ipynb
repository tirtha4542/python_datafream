```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

```python
x= np.array([1,2,3,4])
y = np.array([2,3,5,4])
```

```python
df = pd.DataFrame({'x':x,'y':y})
```

```python
df
```

```python
X = df[['x']]
Y= df['y']
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression().fit(X,Y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y = (0.8*4)+1.5000
y
```

```python
plt.scatter(X,Y)
plt.plot(X,model.predict(X))
plt.show()
```

```python
model.predict([[7]])
```

```python
from sklearn.metrics import mean_squared_error
```

```python
model.score(X,Y)
```
